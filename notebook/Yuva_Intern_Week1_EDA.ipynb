## 8. Missing Value Analysis

Missing values can affect the accuracy of analysis and visualization.

### Missing Value Percentage

The percentage of missing values is calculated to understand the extent of missing data in each column.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

missing_table = missing_table[missing_table["Missing Values"] > 0]

missing_table.sort_values(
    by="Percentage",
    ascending=False
)
```

# 10. Data Cleaning

Data cleaning is performed before visualization and statistical analysis.

The main cleaning operations include:

1. Removing duplicate records
2. Handling missing numerical values
3. Checking categorical values
4. Checking numerical values for invalid or negative values

## 10.1 Remove Duplicate Records

Duplicate rows are removed to ensure that each product record is not counted multiple times.

```python
before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(df)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

## 10.2 Check Numerical Columns

The numerical columns are identified so that missing values and invalid values can be handled appropriately.

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(list(numeric_columns))
```

## 10.3 Handle Missing Numerical Values

For numerical nutritional attributes, missing values are replaced using the median of the respective column.

Median imputation is used because it is less affected by extreme values than the mean.

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values handled.")
```

## 10.4 Verify Missing Values After Cleaning

The dataset is checked again to verify whether missing values remain.

```python
remaining_missing = df.isnull().sum()

print("Remaining missing values:")
print(remaining_missing[remaining_missing > 0])
```

## 11.1 Product Category Distribution

A count plot is used to identify the number of products available in each food category.

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Category",
    order=df["Category"].value_counts().index
)

plt.title("Distribution of Food Products by Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    "../visualizations/univariate/category_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

# 13. Multivariate Analysis

Multivariate analysis examines relationships among multiple variables simultaneously.

This helps identify broader nutritional patterns and correlations within the processed food dataset.

```python
correlation = df[numeric_columns].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()

plt.savefig(
    "../visualizations/multivariate/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## 13.2 Category-wise Nutritional Analysis

Average nutritional values are compared across different food categories.

This helps identify which categories tend to have higher or lower levels of calories, protein, carbohydrates, fat, and sugar.

```python
numeric_columns = [
    "Serving_Size_g",
    "Calories_kcal",
    "Carbohydrates_g",
    "Proteins_g",
    "Total_Fat_g",
    "Saturated_Fat_g",
    "Trans_Fat_g",
    "Sugar_g",
    "Sodium_mg",
    "Dietary_Fiber_g",
    "Cholesterol_mg",
    "Calcium_mg",
    "Iron_mg",
    "Price_INR"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

## 13.3 Average Calories by Food Category

```python
plt.figure(figsize=(10, 6))

category_calories = (
    df.groupby("Category")["Calories_kcal"]
    .mean()
    .sort_values(ascending=False)
)

sns.barplot(
    x=category_calories.values,
    y=category_calories.index
)

plt.title("Average Calories by Food Category")
plt.xlabel("Average Calories (kcal)")
plt.ylabel("Food Category")
plt.tight_layout()

plt.savefig(
    "../visualizations/multivariate/category_average_calories.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
